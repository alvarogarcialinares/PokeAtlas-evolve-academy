# PokéAtlas — Comprobación de Gemini

Este notebook comprueba únicamente tres cosas:

1. que las dependencias están instaladas;
2. que el archivo `.env` contiene `GEMINI_API_KEY`;
3. que podemos hacer una llamada real a Gemini.

> La API key nunca debe escribirse directamente en el notebook ni subirse a GitHub.


## 1. Instalación

Ejecuta esta celda solo si todavía no has instalado las dependencias:

```bash
pip install -U google-genai python-dotenv langchain-google-genai
```


````python
from pathlib import Path
import os
from dotenv import load_dotenv

# El notebook está pensado para vivir dentro de PokeAtlas/notebooks/
env_path = Path("../.env")

if not env_path.exists():
    raise FileNotFoundError(
        "No encuentro ../.env. Crea el archivo .env en la raíz de PokeAtlas "
        "y añade GEMINI_API_KEY=tu_clave"
    )

load_dotenv(env_path)

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError(
        "El archivo .env existe, pero no contiene GEMINI_API_KEY."
    )

print("✅ GEMINI_API_KEY encontrada.")
print("La clave no se muestra por seguridad.")
````

## 2. Probar el SDK oficial de Gemini

````python
from google import genai

client = genai.Client()

# Modelo rápido y adecuado para comprobar la conexión.
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Responde únicamente: Conexión con Gemini correcta."
)

print(response.text)
````

## Resultado esperado

Si todo está bien, deberías ver algo parecido a:

```text
✅ GEMINI_API_KEY encontrada.
La clave no se muestra por seguridad.

Conexión con Gemini correcta.
```

Si aparece un error de modelo, cuota, autenticación o librería, copia el error completo para revisarlo antes de continuar con ChromaDB y LangGraph.
